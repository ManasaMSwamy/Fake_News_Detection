```python
import pandas as pd
import numpy as np

import matplotlib.pyplot as plt
import seaborn as sns

pd.set_option('display.max_columns', None)
```

```python
fake_df = pd.read_csv("Fake.csv")
true_df = pd.read_csv("True.csv")
```

1. Find the total number of news articles.

```python
print("Total News Articles:", len(fake_df))
print("Total News Articles:", len(true_df))
```

2.Find the count of Fake and Real news.

```python
fake_df["label"] = "Fake"
true_df["label"] = "Real"

print(fake_df["label"].value_counts())
print(true_df["label"].value_counts())
```

3. Find the percentage of Fake and Real news

```python
df = pd.concat([fake_df, true_df], ignore_index=True)
print((df["label"].value_counts(normalize=True) * 100).round(2))
```

4. Find the total number of subjects

```python
print("Total Subjects:", df["subject"].nunique())
```

  5. Find all available subjects

```python
print(df["subject"].unique())
```

6. Find the count of news articles in each subject

```python
df["subject"].value_counts()
```

7. Find Fake and Real news count for each subject

```python
pd.crosstab(df["subject"], df["label"])
```

8. Find the average article length

```python
df["text_len"] = df["text"].apply(len)

print("Average Article Length:", df["text_len"].mean())
```

9. Find the average word count

```python
df["word_count"] = df["text"].apply(lambda x: len(x.split()))

print("Average Word Count:", df["word_count"].mean())
```

10. Find the longest news article

```python
df.loc[df["word_count"].idxmax()]
```

11. Find the shortest news article

```python
df.loc[df["word_count"].idxmin(), ["title", "text", "word_count"]]
```

12. Find the top 10 longest news articles

```python
df.nlargest(10, "word_count")[["title", "word_count"]]
```

13. Find the top 10 shortest news articles

```python
df.nsmallest(10, "word_count")[["title", "word_count"]]
```

14. Find the earliest and latest news date

```python
print("Earliest Date:", df["date"].min())
print("Latest Date:", df["date"].max())
```

15. Find the number of news articles published each month

```python
print(df["date"].dtype)
df["date"] = df["date"].astype(str).str.strip()

df["date"] = pd.to_datetime(
    df["date"],
    errors="coerce",
    format="mixed"
)

print(df["date"].dtype)
df.groupby(df["date"].dt.to_period("M")).size()
```

16. Find the month with the highest number of news articles

```python
df.groupby(df["date"].dt.to_period("M")).size().sort_values(ascending=False).head()
```
